```python
import seaborn as sns 
import pandas as pd 
titanic_df=pd.read_csv('Titanic-Dataset.csv')
```

```python
#heat map for correlation between some columns
d_df=titanic_df[['Age','Fare','Pclass','Survived']]
sns.heatmap(d_df.corr(),annot=True,cmap='coolwarm')
```

```python
#box plot for fare distribution by passenger class 
titanic_df['Pclass']=titanic_df['Pclass'].astype(str)
sns.boxplot(data=titanic_df,x='Fare',y='Pclass')
titanic_df['Pclass']=titanic_df['Pclass'].astype(int) #reseting type 
```

```python
#box plot for age distribution by survival status 
titanic_df['Survived']=titanic_df['Survived'].replace({1:'Survivors',0:'Non-Survivors'})
sns.boxplot(titanic_df,x='Age',y='Survived')
titanic_df['Survived']=titanic_df['Survived'].replace({'Survivors':1,'Non-Survivors':0})
```

```python
#histogram for age distribution 
sns.histplot(titanic_df,x='Age',kde=True,hue='Sex')
```

```python
#barplot for survival rate by gender 
sr=titanic_df[titanic_df['Survived']==1]['Sex'].value_counts()/titanic_df['Sex'].value_counts()
sns.barplot(x=sr.index,y=sr.values,hue=sr.index)
```

```python
#pair plot 
sns.pairplot(titanic_df[['Age','Fare','Pclass','Survived']],hue='Survived')
```

```python
#violin plot for age distribution by passenger class 
titanic_df['Pclass']=titanic_df['Pclass'].astype(str)
sns.violinplot(titanic_df,x='Age',y='Pclass')
titanic_df['Pclass']=titanic_df['Pclass'].astype(int)
```

```python
#scatter plot 
sns.scatterplot(titanic_df,x='Age',y='Fare',hue='Survived')
```
